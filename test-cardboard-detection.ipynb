```python
from ultralytics import YOLO
```

```python
model = YOLO('/home/ict/projects/yoloProjects/counting-cardboard/runs/segment/runs/counting-cardboard-v1/weights/best.pt')

for video in os.listdir('detection_test/videos'):
    if video.endswith('.mp4'):
        print(f"Processing video: {video}")
        results = model.predict(
            source=f'detection_test/videos/{video}',
            save=True,
            conf=0.8,
            device=0,
            project='detection_test/results',
            name=video.replace('.mp4', ''),
            stream=True
        )
        for r in results:
            pass
        print(f"Selesai: {video}")
```

```python
import subprocess
import glob
avi_files = glob.glob('runs/segment/detection_test/results/*/*.avi')
for avi in avi_files:
    mp4_path = avi.replace('.avi', '.mp4')
    subprocess.run([
        'ffmpeg', '-i', avi, '-vcodec', 'libx264', '-y', mp4_path
    ])
    print(f"Converted: {mp4_path}")
```

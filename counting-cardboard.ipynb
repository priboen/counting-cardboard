```python
from ultralytics.data.converter import convert_coco
```

```python
for split in ["train", "valid", "test"]:
  convert_coco(
    labels_dir=f"{base}/{split}/",
    save_dir=f"yolo_dataset/labels_temp_{split}/",
    use_segments=True
  )
```

```python
for split in ["train","valid","test"]:
  out_dir=f"yolo_dataset/labels_temp_{split}/labels"
  print(split, "->", os.listdir(out_dir)[:3] if os.path.exists(out_dir) else "folder tidak ditemukan")
```

```python
import shutil, glob
```

```python
for split in ["train", "valid", "test"]:
  os.makedirs(f"yolo_dataset/images/{split}", exist_ok=True)
  os.makedirs(f"yolo_dataset/labels/{split}", exist_ok=True)

  img_files = []
  for ext in ["*.jpg", "*.JPG", "*.jpeg", "*.JPEG", "*.png", "*.PNG"]:
    img_files.extend(glob.glob(f"{base}/{split}/{ext}"))
  for img in img_files:
    shutil.copy(img, f"yolo_dataset/images/{split}/")

  label_src = f"yolo_dataset/labels_temp_{split}/labels/_annotations.coco"
  for lbl in glob.glob(f"{label_src}/*.txt"):
    shutil.copy(lbl, f"yolo_dataset/labels/{split}/")

  print(split, "-> gambar :", len(os.listdir(f"yolo_dataset/images/{split}")), "| label: ", len(os.listdir(f"yolo_dataset/labels/{split}")))
```

```python
from ultralytics import YOLO

model = YOLO('yolo11n-seg.pt')

results = model.train(
  data='data.yaml',
  epochs=100,
  imgsz=640,
  batch=16,
  patience=20,
  device=0,
  mosaic=1.0,
  mixup=0.1,
  copy_paste=0.3,
  project='runs',
  name='counting-cardboard-v1'
)
```

```python
metrics = model.val(data='data.yaml', split='test')
print("Box mAP50-95:", metrics.box.map)
print("Mask mAP50-95:", metrics.seg.map)
```
